# Targeted audits + multimodel forgetting
One quiet resumable worker. Extract all Python files together. Run Setup → Launch; refresh manually. The whole expanded queue may require multiple sessions. See README.md for the fixed prediction protocol and numerical limits.

```python
from pathlib import Path
import paper_suite as study

OLMO_SOURCE = Path('/home/ubuntu/1/runs/olmo_association_v1')
FOUR_VECTOR_RESULTS = Path('/home/ubuntu/4/runs/olmo_precise_forgetting_v1/runs/olmo_four_vector_v1')
# FOUR_VECTOR_RESULTS can instead point to your four_vector_share.zip.
LAST_OUTPUT = Path.cwd() / '.paper_suite_last_output.txt'
OUTPUT = Path(LAST_OUTPUT.read_text().strip()) if LAST_OUTPUT.exists() else Path.cwd() / 'runs' / 'multimodel_paper_v1'
saved = study.f.read(OUTPUT / 'settings.json')
SETTINGS = saved if saved and saved['olmo_source'] == str(OLMO_SOURCE.resolve()) and saved['previous'] == str(FOUR_VECTOR_RESULTS.resolve()) else study.defaults(OLMO_SOURCE, FOUR_VECTOR_RESULTS, OUTPUT)
SETTINGS['hours'] = 11.5
# Optional before the first launch: disable Qwen to prioritize the two Pythias.
# SETTINGS['models'][2]['enabled'] = False
```

## Launch or resume
Model revisions and original-OLMo predictor fits are frozen before new-model results are collected.

```python
OUTPUT = Path(study.launch(SETTINGS))
_ = LAST_OUTPUT.write_text(str(OUTPUT))
```

## Refresh on demand

```python
_ = study.refresh(OUTPUT)
```
