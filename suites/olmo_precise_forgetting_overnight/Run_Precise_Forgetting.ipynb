## Controls
Run All merely defines these functions. Call `stop_run()` in a new cell to pause, then wait for `alive=False`. Resume with Start. Call `fresh_run()` only to create a separate fresh run. Call `export_results()` for a share ZIP.

```python
def stop_run():
    study.stop(SETTINGS["output"])

def fresh_run():
    return study.restart(SETTINGS)

def export_results():
    path = study.export(SETTINGS["output"])
    print("Share ZIP:", path)
    return path
```
