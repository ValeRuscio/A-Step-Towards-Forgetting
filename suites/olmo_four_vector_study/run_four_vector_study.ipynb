# Four-vector natural forgetting study
Run Setup, then Launch. The worker runs quietly in the background. Re-run Refresh to check it; Show results prints numbers and plots only when requested. All Python files must stay together. See README.md for definitions, checks and limitations.


```python
from pathlib import Path
import four_vector_study as study

SOURCE = Path('/home/ubuntu/1/runs/olmo_association_v1')
LAST_OUTPUT = Path.cwd() / '.four_vector_last_output.txt'
OUTPUT = Path(LAST_OUTPUT.read_text().strip()) if LAST_OUTPUT.exists() else Path.cwd() / 'runs' / 'olmo_four_vector_v1'
saved = study.read(OUTPUT / 'settings.json')
SETTINGS = saved if saved and Path(saved['source']) == SOURCE.resolve() else study.defaults(SOURCE, OUTPUT)
SETTINGS['hours'] = 11.5
# Full population gradients, all natural steps, and selected finite paths are the defaults.
# SETTINGS['curvature_points_per_event'] = 2  # More HVP sites; more time.
```

## Launch or resume
Safe to re-run: an active worker is left running, and a paused worker resumes completed records.

```python
OUTPUT = Path(study.launch(SETTINGS))
_ = LAST_OUTPUT.write_text(str(OUTPUT))
```

## Refresh on demand
This is a snapshot, not a continuously refreshing display. No output changes until you re-run the cell.

```python
_ = study.refresh(OUTPUT)
```

## Show saved numbers and plots
Can be used while the worker is running.

```python
study.show(OUTPUT)
```

## Stop when needed
Requests a cooperative stop. Wait until Refresh reports `alive=False`, then use Launch to resume.

```python
STOP_NOW = False  # Change to True only when you want to stop.
if STOP_NOW:
    study.stop(OUTPUT)
```

## Export all saved measurements
Creates a share ZIP inside OUTPUT. The download link appears below. It contains metrics and code, not model checkpoints.

```python
from IPython.display import display, FileLink
archive = Path(study.export(OUTPUT))
print(archive)
display(FileLink(str(archive.relative_to(Path.cwd())) if archive.is_relative_to(Path.cwd()) else str(archive)))
```
